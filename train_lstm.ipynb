```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense,Dropout
from tensorflow.keras.callbacks import ModelCheckpoint
# This saves the best version of the model during training
```

## Loading processed data

```python
# Loading the prepared datasets from processed_data/
X_lstm=np.load("processed_data/X_lstm.npy")
Y_lstm=np.load("processed_data/Y_lstm.npy")
```

```python
print("X_lstm shape:", X_lstm.shape)   
print("Y_lstm shape:", Y_lstm.shape)
```

## Building the LSTM model

```python
model=Sequential()
model.add(LSTM(64, return_sequences=False, input_shape=(20, 5)))
model.add(Dropout(0.2))
model.add(Dense(32, activation='relu'))   
model.add(Dense(2, activation='linear'))
```

```python
model.compile(optimizer="adam",loss='mse')
model.summary()
```

## Creating a model saver

```python
checkpoint = ModelCheckpoint(
    "best_lstm.h5",
    monitor="loss",
    save_best_only=True
)
```

## Training the LSTM

```python
history = model.fit(X_lstm,Y_lstm,epochs=100,batch_size=32,shuffle=True,callbacks=[checkpoint])
# batch_size --> This controls how many samples are trained at once
# shuffle --> This randomizes the training samples each epoch
# callbacks --> This saves the best version of your model automatically.
```

## Saving the model

```python
model.save("final_lstm_model.keras")
```
