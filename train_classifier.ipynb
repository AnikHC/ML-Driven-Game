```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.callbacks import ModelCheckpoint
```

## Loading processed classifier data

```python
X_cls=np.load("processed_data/X_cls.npy")
Y_cls=np.load("processed_data/Y_cls.npy")
```

```python
print("X_cls shape:",X_cls.shape)
print("Y_cls shape:",Y_cls.shape)
```

## One-hot encoding labels

```python
Y_cls_oh=tf.keras.utils.to_categorical(Y_cls,num_classes=4)
# 0 = idle
# 1 = attack
# 2 = dash
# 3 = move
```

## Building the model

```python
model=Sequential()
model.add(Dense(64,activation='relu',input_shape=(5,)))
model.add(Dropout(0.1))
model.add(Dense(64,activation='relu'))
model.add(Dropout(0.1))
model.add(Dense(4,activation='softmax')) # for multiclass classification
```

```python
model.compile(loss='categorical_crossentropy',optimizer='adam',metrics=['accuracy'])
model.summary()
```

## Setting up model checkpoint

```python
checkpoint = ModelCheckpoint(
    "best_classifier.h5",
    monitor="accuracy",
    save_best_only=True
)
```

## Training the classifier

```python
history = model.fit(X_cls,Y_cls_oh,epochs=40,    batch_size=32,shuffle=True,callbacks=[checkpoint])
```

```python
model.save("final_classifier_model.keras")
```
