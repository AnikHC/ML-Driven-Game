```python
import tensorflow as tf
import tf2onnx
```

## Loading trained model

```python
model=tf.keras.models.load_model("best_lstm.h5")
```

## Converting model --> ONNX

```python
# Fix the input shape so Unity Barracuda understands it
input_signature = (
    tf.TensorSpec((1, 20, 5), tf.float32, name="lstm_input"),
)
```

```python
# Convert the model to ONNX with fixed input dimensions
tf2onnx.convert.from_keras(
    model,
    input_signature=input_signature,
    opset=13,
    output_path="lstm_predict_fixed.onnx"
)
```
